```python
spark.sql("USE CATALOG tfmlakehouse")
```

```python
def build_tables(schema: str) -> None:
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS {schema}")

    table_list = dbutils.fs.ls(f"abfss://lakehouse@tfmmbtasta.dfs.core.windows.net/{schema}/")

    for table in table_list:
        table_name = table.name.replace("/", "")
        spark.sql(f"""
        CREATE TABLE tfmlakehouse.{schema}.{table}
        USING DELTA
        LOCATION 'abfss://lakehouse@tfmmbtasta.dfs.core.windows.net/{schema}/{table}/'
        """)

        print(f"Table: {table} builded from route")
```

```python
build_tables("bronze")
build_tables("silver")
build_tables("gold")
```
